```python
import numpy as np
import pandas as pd
```

```python
ecomm_df = pd.read_csv("/content/drive/MyDrive/Asistente IA Data/ECOMM DATA.xlsx - Orders.csv")
```

```python
ecomm_df.columns
```

```python
ecomm_df_sales = ecomm_df[['Order ID','Order Date','Customer ID','Product ID','Sales','Quantity','Profit']]

ecomm_df_clients = ecomm_df[['Customer ID','Customer Name','City','State','Country']]
ecomm_df_clients_uniques = ecomm_df_clients.drop_duplicates(subset = ["Customer ID"], keep="first")

ecomm_df_products = ecomm_df[['Product ID','Product Name','Category','Sub-Category']]
ecomm_df_products_uniques = ecomm_df_products.drop_duplicates(subset = ["Product ID"], keep="first")
```

```python
ecomm_df_sales.to_csv('ECOMM Sales.csv', index=False)
ecomm_df_clients_uniques.to_csv('ECOMM Clients.csv', index=False)
ecomm_df_products_uniques.to_csv('ECOMM Products.csv', index=False)

print('Listo')
```

```python
retail_df = pd.read_csv("/content/new_retail_data.csv")
```

```python
retail_df.columns
```

```python
retail_df_sales = retail_df[['Transaction_ID','Date','Customer_ID','Amount','Total_Purchases','Total_Amount']]
retail_df_clients = retail_df[['Customer_ID', 'Name','Email','Phone','City','State','Country','Age','Gender']]
retail_df_products = retail_df[['products','Product_Category','Product_Type']]
```

```python
retail_df_sales.columns = ['Order ID','Order Date','Customer ID','Sales','Quantity','Total_Amount']

retail_df_clients.columns = ['Customer ID', 'Customer Name','Email','Phone','City','State','Country','Age','Gender']
retail_df_clients_uniques = retail_df_clients.drop_duplicates(subset = ["Customer ID"], keep="first")

retail_df_products.columns = ['Product Name','Category','Sub-Category']
retail_df_products_uniques = retail_df_products.drop_duplicates(subset = ["Product Name"], keep="first")
```

```python
retail_df_sales.to_csv('Retail Sales.csv', index=False)
retail_df_clients_uniques.to_csv('Retail Clients.csv', index=False)
retail_df_products_uniques.to_csv('Retail Products.csv', index=False)
print('Listo')
```
