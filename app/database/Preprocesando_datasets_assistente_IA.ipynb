```python
retail_df = pd.read_csv("/content/drive/MyDrive/Asistente IA Data/new_retail_data.csv")
retail_df.head()
```

```python
retail_df_products = retail_df[['products','Product_Category','Product_Type','Product_Brand']]
retail_df_products.columns = ['Product Name','Category','Sub-Category','Product_Brand']
retail_df_products_uniques = retail_df_products.drop_duplicates(subset = ['Product Name','Category','Sub-Category','Product_Brand'], keep='first')
```

```python
# Rellenando los Nan en el DataFrame de products
brand_category_mapping = {
    'Samsung': 'Electronics',
    'Apple': 'Electronics',
    'Sony': 'Electronics',
    'Nestle': 'Grocery',
    'Pepsi': 'Grocery',
    'Coca-Cola': 'Grocery',
    'HarperCollins': 'Books',
    'Random House': 'Books',
    'Penguin Books': 'Books',
    'Adidas': 'Clothing',
    'Zara': 'Clothing',
    'Nike': 'Clothing',
    'Home Depot': 'Home Decor',
    'Bed Bath & Beyond': 'Home Decor',
    'IKEA': 'Home Decor'
}


for brand, category in brand_category_mapping.items():
    retail_df_products_uniques.loc[retail_df_products_uniques['Product_Brand'] == brand, 'Category'] = category

for category in retail_df_products_uniques['Category'].unique():
    retail_df_products_uniques.loc[(retail_df_products_uniques['Category'] == category) & (retail_df_products_uniques['Product_Brand'].isnull()), 'Product_Brand'] = category

```

```python
retail_df_products_uniques = retail_df_products_uniques.drop_duplicates(subset = ['Product Name','Category','Sub-Category','Product_Brand'], keep='first')
retail_df_products_uniques
```

```python
num_productos = retail_df_products_uniques.shape[0]
product_ids = range(1, num_productos + 1)
retail_df_products_uniques['Product ID'] = product_ids
```

```python
# Ordenando el dataframe de products
cols = retail_df_products_uniques.columns.tolist()
cols = cols[-1:] + cols[:-1]
retail_df_products_uniques = retail_df_products_uniques[cols]
```

```python
retail_df_products_uniques
```

```python
retail_df_sales = retail_df[['Transaction_ID','Date','Customer_ID','Amount','Total_Purchases','Total_Amount','products','Product_Category','Product_Type','Product_Brand']]
retail_df_sales.columns = ['Order ID','Order Date','Customer ID','Sales','Quantity','Total_Amount','Product Name','Category','Sub-Category','Product_Brand']
retail_df_sales = retail_df_sales.dropna()

retail_df_sales['Order ID'] = retail_df_sales['Order ID'].astype(int)
retail_df_sales['Customer ID'] = retail_df_sales['Customer ID'].astype(int)
```

```python
retail_df_sales
```

```python
# Rellenando los Nan en el DataFrame de products
brand_category_mapping = {
    'Samsung': 'Electronics',
    'Apple': 'Electronics',
    'Sony': 'Electronics',
    'Nestle': 'Grocery',
    'Pepsi': 'Grocery',
    'Coca-Cola': 'Grocery',
    'HarperCollins': 'Books',
    'Random House': 'Books',
    'Penguin Books': 'Books',
    'Adidas': 'Clothing',
    'Zara': 'Clothing',
    'Nike': 'Clothing',
    'Home Depot': 'Home Decor',
    'Bed Bath & Beyond': 'Home Decor',
    'IKEA': 'Home Decor'
}


for brand, category in brand_category_mapping.items():
    retail_df_sales.loc[retail_df_sales['Product_Brand'] == brand, 'Category'] = category

for category in retail_df_sales['Category'].unique():
    retail_df_sales.loc[(retail_df_sales['Category'] == category) & (retail_df_sales['Product_Brand'].isnull()), 'Product_Brand'] = category

```

```python
retail_df_sales = retail_df_sales.merge(retail_df_products_uniques[['Product ID', 'Product Name', 'Category', 'Sub-Category', 'Product_Brand']],
                                               on=['Product Name', 'Category', 'Sub-Category', 'Product_Brand'],
                                               how='left')
```
